```python
df_recent = df[df['기준연도'].isin([2022, 2023, 2024, 2025, 2026])]
df_recent.head(5)
```

```python
df_recent.groupby("학교명")["모집인원_정원내"].sum()
```

```python
# 본교와 학교명이 다른 이원화캠퍼스를 가진 대학교는 학습 데이터에서 제거
exclude_prefixes = (
    '가야대학교',
    '영산대학교',
    '목포대학교',
    '국립목포대학교',
    '창원대학교',
    '국립창원대학교',
    '강원대학교',
    '강릉원주대학교',
    '국립강릉원주대학교',
)

df_recent = df_recent[~df_recent['학교명'].str.startswith(exclude_prefixes, na=False)].copy()
```

```python
df_recent[~df_recent['본분교명'].isin(['본교', '분교'])]
```

```python
# df2, 이원화캠퍼스가 존재하는 학교명 시리즈
df2 = df_recent[~df_recent['본분교명'].isin(['본교', '분교'])]["학교명"].drop_duplicates()
df2.head(5)
```

```python
# df3, 이원화캠퍼스, 분교를 제외한 본교의 데이터테이블
df3 = df_recent[df_recent["본분교명"] == "본교"]
print(df3.head(5))
print(df3.shape)
```

```python
# df2에 해당하는 학교명을 가진 행의 학교명과 학교코드 열만 추출
# df4, 이게 본교 학교코드 데이터테이블
df4 = df3[df3["학교명"].isin(df2)][["학교명", "학교코드"]].drop_duplicates()
df4
```

```python
# 그러면 이제 df_recent에서 각 행의 학교명을 df4에 있는 학교명과 비교해서 일치하면
# 이원화캠퍼스(df2)의 학교코드를 df4의 학교코드로 덮어쓰기

df_recent[(df_recent["학교명"].isin(df4["학교명"])) & (~df_recent["본분교명"].isin(["본교", "분교"]))]["학교코드"]
```

```python
# df4는 학교명마다 학교코드가 하나씩 있어야 함
code_map = df4.set_index('학교명')['학교코드']

# 학교명에 해당하는 df4의 학교코드 조회
mapped_codes = df_recent['학교명'].map(code_map)

# 학교명이 일치하고, 분교가 아닌 행 선택
mask = mapped_codes.notna() & df_recent['본분교명'].ne('분교')

# 선택된 행의 학교코드를 덮어쓰기
df_recent.loc[mask, '학교코드'] = mapped_codes.loc[mask]
df_recent[df_recent["본분교명"] == "제2캠퍼스"]
```
